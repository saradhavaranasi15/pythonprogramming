```python
# to fnd no.of duplicates in a list
my_list = [1, 2, 2, 3, 4, 4, 4, 5]
seen = []
duplicate_count = 0

for item in my_list:
  if item in seen:
    duplicate_count += 1
  else:
    seen.append(item)

print("Number of duplicates:", duplicate_count)  # Output: 3

```

```python
# program to find a key exists in dictionary or not
student = {"name": "Rahul", "age": 20, "course": "Python"}

key_to_check = "age"

if key_to_check in student:
  print("Key exists! Value is:", student[key_to_check])
else:
  print("Key does not exist.")

```
