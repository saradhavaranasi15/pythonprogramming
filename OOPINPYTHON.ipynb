```python

class shape:
    def area(self):
        pass
    def perimeter(self):
        pass

class circle(shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return 3.14 * self.r * self.r
    def perimeter(self):
        return 2 * 3.14 * self.r

class square(shape):
    def __init__(self, side):
        self.side = side
    def area(self):
        return self.side ** 2
    def perimeter(self):
        return 4 * self.side

class triangle(shape):
    def __init__(self, a, b, c, h):
        self.a = a
        self.b = b
        self.c = c
        self.h = h
    def area(self):
        return 0.5 * self.b * self.h
    def perimeter(self):
        return self.a + self.b + self.c

c1 = circle(4)
print("Circle area :", c1.area())
print("Circle perimeter:", c1.perimeter())

d1 = square(6)
print("Square area:", d1.area())
print("Square perimeter:", d1.perimeter())

s1 = triangle(3, 4, 5, 6)
print("Triangle area:", s1.area())
print("Triangle perimeter:", s1.perimeter())


```
