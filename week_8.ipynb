```python
import pandas as pd

data = {
    'Name': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J'],
    'Age': list(range(21, 31)),
    'Marks': list(range(80, 90)),
    'Salary': list(range(30000, 40000, 1000)),
    'Grade': ['A', 'A', 'A', 'B', 'A', 'A', 'B', 'A', 'A', 'A']
}

df = pd.DataFrame(data)

print("Dataframe:")
print(df)

print("\nFirst 5 rows:")
print(df.head())

print("\nSelect Name column:")
print(df['Name'])

print("\nSelect multiple columns:")
print(df[['Name', 'Marks']])

print("\nSelect rows:")
print(df.iloc[0:5])
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Sample data (replace these lists with your actual numbers)
data = {"Age": [10, 12, 14, 16, 18], "Marks": [65, 70, 75, 85, 90]}

df = pd.DataFrame(data)

# Scatter plot
plt.figure()
plt.scatter(df["Age"], df["Marks"])
plt.xlabel("Age")
plt.ylabel("Marks")
plt.title("Age vs Marks - Scatter Plot")
plt.show()

# Line plot
plt.figure()
plt.plot(df["Age"], df["Marks"], marker="o")
plt.xlabel("Age")
plt.ylabel("Marks")
plt.title("Age vs Marks - Line Plot")
plt.show()
```
