OPERATIONS ON LIST
1.Indexing
2.Slicing
3.Concatenation
4.Repetition
5.Membership Operator
6.Iteration

```python
#INDEXING
l=[45,23,"BTS","gpt",77.78,123,67]
print(l[3])
```

```python
#SLICING
l=[45,23,"BTS","gpt",77.78,123,67]
print(l[2:5])
print(l[0:7:2])
print(l[:5])
print(l[0:])
print(l[-1])
print([l[-5:-1]])
print(l[2:8:2])
print(l[::-1])
print(l[::-2])
```

```python
#CONCATENATION
l1=[45,23,"BTS"]
l2=["gpt",77.78,123,67]
print(l1+l2)
```

```python
#REPETITION
print(l*3)
```

```python
#MEMBERSHIP OPERATOR
print(23 in l)
```

```python
#ITERATION
for i in l:
  print(i,end=" ")
```

Program: Print the sum of list elements without using sum() function

```python
L=list(map(int,input().split()))
sum=0
for i in L:
  sum=sum+i
print(sum)
```

Creating a list in run time
1.Using map function
2.List comprehension
3.Normal approach

```python
#NORMAL APPROACH WITHOUT FUNCTIONS
a=[]
n=int(input())
for i in range(n):
  a.append(int(input()))
print(a)
```

```python
#USING MAP FUNCTION
a=list(map(int,input().split())) #print list
a=map(int,input().split()) #returns the address of the location where list is stored
a=input().split() #Strings are stored
a=list(map(int,input().split("/"))) #used when we need to seperate with any comma or slash or any symbol
```

```python
#LIST COMPREHENSION:It is a way of creating a new list using existing sequence
a=[int(i) for i in input().split()]
print(a)
L1=[value for value in range(10)]
print(L1)
L2=[i**2 for i in L1]
print(L2)
L3=[i for  i in L2 if (i%2==0)]
print(L3)
```
